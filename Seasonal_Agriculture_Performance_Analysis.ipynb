## SEASONAL AGRICULTURE PERFORMANCE ANALYSIS

**Import Libraries**

```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)

# Set visualization style
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

**Load the dataset**

```python
# Load the dataset

df = pd.read_csv('/content/seasonal_agriculture_performance_dataset.csv')

print("Dataset loaded successfully!")
```

**Top 5 Rows Analysis**

```python
df.head()
```

**Dataset Shape & Structure**

```python
# Check the shape of the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nDataset structure:")
df.info()
```

**Examine Data Types**

```python
# Check data types of all columns

print("Data types of each column:")
print(df.dtypes)

print("\nSummary of data types:")
print(df.dtypes.value_counts())
```

**Missing Values Identification & Handling**

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])
```

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage[missing_percentage > 0])
```

**Handle Missing Values**

```python
# Handle missing values using median

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

print("Missing values handled successfully!")
```

```python
# Verify that no missing values remain

print("Total missing values:", df.isnull().sum().sum())
```

**Duplicate Records**

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Handle duplicate records

df = df.drop_duplicates()

print("Duplicate records handled successfully!")
print("New dataset shape:", df.shape)
```

**Descriptive Statistical Analysis**

```python
# Descriptive statistical analysis

df.describe()
```

```python
# Descriptive analysis of categorical columns

df.describe(include='object')
```

**Outlier Investigation**

```python
# Select important numerical columns for outlier investigation

outlier_columns = [
    'Farm_Area_Hectares',
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Pesticide_Litre_ha',
    'Yield_Tonnes_Ha',
    'Profit_INR',
    'Water_Used_m3',
    'Disease_Pest_Risk_pct'
]

# Create boxplots
for column in outlier_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f'Outlier Investigation - {column}')
    plt.xlabel(column)
    plt.show()
```

**Univariate Analysis**

```python
# Crop distribution

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Crop')
plt.title('Distribution of Crops')
plt.xlabel('Crop')
plt.ylabel('Number of Farms')
plt.xticks(rotation=45)
plt.show()
```

```python
# Season distribution

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Season')
plt.title('Distribution of Agricultural Seasons')
plt.xlabel('Season')
plt.ylabel('Number of Farms')
plt.show()
```

```python
# Yield distribution

plt.figure(figsize=(8, 5))
sns.histplot(df['Yield_Tonnes_Ha'], kde=True)
plt.title('Distribution of Crop Yield')
plt.xlabel('Yield (Tonnes per Hectare)')
plt.ylabel('Frequency')
plt.show()
```

**Bivariate Analysis**

```python
# Rainfall vs Yield

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Rainfall_mm', y='Yield_Tonnes_Ha')
plt.title('Rainfall vs Crop Yield')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

```python
# Fertilizer vs Yield

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Fertilizer_kg_ha', y='Yield_Tonnes_Ha')
plt.title('Fertilizer Usage vs Crop Yield')
plt.xlabel('Fertilizer (kg/ha)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

```python
# Season vs Yield

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Season', y='Yield_Tonnes_Ha')
plt.title('Yield Performance Across Seasons')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

```python
# Irrigation Method vs Yield

plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='Irrigation_Method', y='Yield_Tonnes_Ha')
plt.title('Yield Performance by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()
```

**Multivariate Analysis**

```python
# Multivariate analysis

multi_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Soil_Moisture_pct',
    'Fertilizer_kg_ha',
    'Yield_Tonnes_Ha'
]

sns.pairplot(df[multi_columns])
plt.show()
```

```python
# Average yield by Season and Crop

season_crop_yield = df.groupby(
    ['Season', 'Crop']
)['Yield_Tonnes_Ha'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(
    data=season_crop_yield,
    x='Crop',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title('Average Yield by Crop and Season')
plt.xlabel('Crop')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()
```

**Correlation Analysis**

```python
# Correlation analysis

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Matrix of Agricultural Variables')
plt.show()
```

```python
# Correlation of variables with crop yield

yield_correlation = correlation_matrix['Yield_Tonnes_Ha'].sort_values(
    ascending=False
)

print(yield_correlation)
```

**Seasonal Comparisons**

```python
# Average yield by season

season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)

print(season_yield)

plt.figure(figsize=(8, 5))
sns.barplot(
    x=season_yield.index,
    y=season_yield.values
)

plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.show()
```

```python
# Average profit by season

season_profit = df.groupby('Season')['Profit_INR'].mean().sort_values(ascending=False)

print(season_profit)

plt.figure(figsize=(8, 5))
sns.barplot(
    x=season_profit.index,
    y=season_profit.values
)

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
plt.show()
```

```python
# Seasonal performance comparison

season_summary = df.groupby('Season').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Profit_INR': 'mean',
    'Rainfall_mm': 'mean',
    'Disease_Pest_Risk_pct': 'mean'
}).round(2)

print(season_summary)
```

**Student-Designed Analysis**

```python
# Student-Designed Analysis 1
# Average profit by crop

crop_profit = df.groupby('Crop')['Profit_INR'].mean().sort_values(ascending=False)

print("Average Profit by Crop:")
print(crop_profit.round(2))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=crop_profit.index,
    y=crop_profit.values
)

plt.title('Average Profit by Crop')
plt.xlabel('Crop')
plt.ylabel('Average Profit (INR)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Student-Designed Analysis 2
# Average water efficiency by irrigation method

irrigation_efficiency = df.groupby(
    'Irrigation_Method'
)['Water_Efficiency_t_per_1000m3'].mean().sort_values(ascending=False)

print("Average Water Efficiency by Irrigation Method:")
print(irrigation_efficiency.round(2))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=irrigation_efficiency.index,
    y=irrigation_efficiency.values
)

plt.title('Water Efficiency by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Water Efficiency (Tonnes per 1000 m³)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Student-Designed Analysis 3
# Average yield by state

state_yield = df.groupby('State')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)

print("Average Yield by State:")
print(state_yield.round(2))
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=state_yield.index,
    y=state_yield.values
)

plt.title('Average Crop Yield by State')
plt.xlabel('State')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()
```

## Meaningful Insights

1. **Kharif was the best-performing season:** Kharif recorded the highest average yield of **5.63 tonnes/ha**, compared with 5.04 tonnes/ha in Rabi and 4.64 tonnes/ha in Zaid.

2. **Kharif generated the highest profit:** The average profit during Kharif was **₹178,914.65**, considerably higher than Rabi at ₹87,689.47 and Zaid at -₹24,804.82.

3. **Rainfall varied significantly by season:** Kharif had the highest average rainfall of **849.20 mm**, followed by Rabi at 437.62 mm and Zaid at 304.65 mm.

4. **Sugarcane was the most profitable crop:** Sugarcane recorded the highest average profit of approximately **₹817,188**, followed by Chilli at approximately ₹750,878.

5. **Some crops showed negative profitability:** Wheat, Rice, Maize, and Pulses recorded negative average profits, indicating a need to examine their production costs, yields, and market prices.

6. **Water efficiency differed by irrigation method:** Rainfed farming recorded the highest measured water-efficiency value of **7.56 tonnes per 1000 m³**, followed by Drip at 6.27. This metric should be interpreted carefully because rainfed farming depends on natural rainfall.

7. **Punjab recorded the highest average yield:** Punjab had the highest average yield at **6.12 tonnes/ha**, followed by Karnataka at 5.69 tonnes/ha and Gujarat at 5.66 tonnes/ha.

8. **Andhra Pradesh recorded the lowest average yield:** Andhra Pradesh had an average yield of **4.63 tonnes/ha** among the states analyzed, suggesting that regional factors could be investigated further.

9. **Kharif had the highest disease and pest risk:** The average disease and pest risk during Kharif was **54.47%**, higher than Rabi and Zaid.

10. **Agricultural performance depends on multiple factors:** The analysis indicates that crop performance is influenced by a combination of season, rainfall, crop type, irrigation, soil conditions, and regional factors rather than a single variable.


## Evidence-Based Recommendations

1. **Focus on Kharif-season farming:** Kharif recorded the highest average yield of 5.63 tonnes/ha and the highest average profit of ₹178,914.65. Therefore, suitable Kharif crops can be prioritized where conditions permit.

2. **Promote profitable crops:** Sugarcane and Chilli recorded the highest average profits of approximately ₹817,188 and ₹750,878 respectively. Farmers can consider these crops based on local soil, climate, water availability, and market conditions.

3. **Review low-profit crops:** Wheat, Rice, and Maize recorded negative average profits in the dataset. Their production costs, yields, and market prices should be reviewed to identify opportunities for improving profitability.

4. **Improve irrigation efficiency:** Flood irrigation recorded the lowest measured water-efficiency value of 3.44 tonnes per 1000 m³, while Drip recorded 6.27. Efficient water-management practices should therefore be evaluated to reduce water wastage.

5. **Strengthen pest and disease management:** Kharif had the highest average disease and pest risk of 54.47%. Additional monitoring and preventive crop-protection measures may help reduce potential losses.

6. **Study high-performing states:** Punjab recorded the highest average yield of 6.12 tonnes/ha. The agricultural practices and conditions associated with higher performance can be studied for useful lessons.

7. **Investigate lower-performing regions:** Andhra Pradesh recorded the lowest average yield of 4.63 tonnes/ha among the states in the dataset. Factors such as crop selection, soil conditions, irrigation, and farming practices should be investigated.

8. **Use seasonal conditions for planning:** Rainfall, temperature, and seasonal conditions should be considered when selecting crops and planning agricultural resources.


## Limitations

1. The analysis is based only on the available dataset and may not represent all agricultural regions.

2. Missing values were handled using median imputation, which may slightly affect the analysis results.

3. Outliers were investigated but not removed because they may represent genuine agricultural conditions.

4. Correlation between variables does not necessarily mean that one variable causes another.

5. The dataset does not include some factors such as farmer experience, government policies, and changing market demand.

6. Seasonal performance may vary across different regions and crops.

7. Profitability depends on the costs and market prices recorded in the dataset and may change over time.




## Final Conclusion

The Seasonal Agriculture Performance Analysis provides useful insights into crop yield, profitability, seasonal performance, irrigation, rainfall, and regional agricultural performance.

Kharif emerged as the best-performing season, recording the highest average yield and profit. Sugarcane and Chilli were the most profitable crops, while Wheat, Rice, and Maize showed negative average profitability.
